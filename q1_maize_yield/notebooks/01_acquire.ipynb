```python

### 2.1 Data Requirements Statement

The MCSU requires a dataset that can support the preparation of maize yield forecasts for farmer cooperatives in Rwanda.

Target variable: Maize yield per hectare. The supplied CSV uses `yield_t_ha`, measured in tonnes per hectare. The assignment document describes `yield_kg_per_ha`, measured in kilograms per hectare. This discrepancy will be reported and confirmed.

Expected input features:
- Farm location: province and district.
- Agricultural information: year, season, seed variety and farming practice.
- Environmental information: altitude, rainfall, average temperature and soil properties.
- Farm inputs: NPK fertiliser, urea, manure and plant density.
- Farm conditions: farm area, irrigation and pest incidence.

Geographical coverage: Rwanda.

Time coverage: The supplied CSV contains records from 2015 to 2024. The assignment scenario describes data collected between 2023 and 2025.

Required records: The assignment describes approximately 1,200 records and 20 columns. The supplied CSV contains 1,426 records and 20 columns.

Data quality requirements: The dataset should be checked for missing values, duplicate records, incorrect data types, inconsistent categories, unrealistic values and possible data leakage.

Expected output: An inspected and documented dataset that can be cleaned and prepared for later modelling.

Modelling restriction: No prediction model will be trained during this assignment.
```

```python

# Import the required libraries
import pandas as pd
import numpy as np
from pathlib import Path

# Define the project directory
project_dir = Path(r"D:\MYProject\ML_Element1")

# Define the raw dataset location
data_path = (
    project_dir
    / "q1_maize_yield"
    / "data"
    / "raw"
    / "maize_yield_rwanda_rw.csv"
)

# Load the original dataset
df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("File:", data_path.name)
```

```python

# Display the number of rows and columns
print("Number of records:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)
```

```python

# Inspect the unique values in categorical columns

categorical_columns = df.select_dtypes(
    include=["object"]
).columns

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

```python

# Prepare the Data Dictionary

dictionary_info = [
    ["record_id", "Unique farm record identifier", "Categorical", "None", "Unique record code"],
    ["year", "Year of agricultural record", "Integer", "Year", "2015-2024 observed"],
    ["season", "Agricultural season", "Categorical", "None", "A, B or C observed"],
    ["province", "Province where the farm is located", "Categorical", "None", "Rwandan provinces observed"],
    ["district", "District where the farm is located", "Categorical", "None", "District names observed"],
    ["altitude_m", "Farm altitude", "Integer", "Metres", "Numerical altitude"],
    ["rainfall_mm", "Rainfall recorded for the farm/season", "Numerical", "Millimetres", "Missing values present"],
    ["avg_temp_c", "Average temperature", "Numerical", "Degrees Celsius", "Numerical temperature"],
    ["soil_ph", "Soil acidity or alkalinity", "Numerical", "pH", "Missing values present"],
    ["soil_type", "Type of soil", "Categorical", "None", "Soil categories observed"],
    ["seed_variety", "Maize seed variety used", "Categorical", "None", "Seed varieties observed"],
    ["farming_practice", "Farming practice used", "Categorical", "None", "Observed practice categories"],
    ["npk_kg_ha", "NPK fertiliser application rate", "Numerical", "kg/ha", "Numerical rate"],
    ["urea_kg_ha", "Urea fertiliser application rate", "Numerical", "kg/ha", "Numerical rate"],
    ["manure_t_ha", "Manure application rate", "Numerical", "tonnes/ha", "Numerical rate"],
    ["plant_density_per_ha", "Plant density per hectare", "Integer", "Plants/ha", "Numerical count"],
    ["farm_area_ha", "Farm area", "Numerical", "Hectares", "Numerical area"],
    ["irrigation", "Whether irrigation was used", "Categorical", "Yes/No", "Observed categories"],
    ["pest_incidence_index", "Numerical pest incidence indicator", "Numerical", "Not specified", "Scale requires confirmation"],
    ["yield_t_ha", "Maize yield per hectare", "Numerical", "Tonnes/ha", "Target variable in supplied CSV"]
]

data_dictionary = pd.DataFrame(
    dictionary_info,
    columns=[
        "column_name",
        "meaning",
        "data_type",
        "unit",
        "allowed_values_or_notes"
    ]
)

display(data_dictionary)
```

```python

# Save the Data Dictionary in the documentation folder

dictionary_path = (
    project_dir
    / "q1_maize_yield"
    / "docs"
    / "data_dictionary.csv"
)

data_dictionary.to_csv(dictionary_path, index=False)

print("Data Dictionary saved successfully!")
print(dictionary_path)
```
