```python

import pandas as pd
from pathlib import Path

project_dir = Path(r"D:\MYProject\ML_Element1")

processed_dir = (
    project_dir
    / "q2_own_scenario"
    / "data"
    / "processed"
)

train_path = processed_dir / "q2_train_final.csv"
validation_path = processed_dir / "q2_validation_final.csv"
test_path = processed_dir / "q2_test_final.csv"

train_check = pd.read_csv(train_path)
validation_check = pd.read_csv(validation_path)
test_check = pd.read_csv(test_path)

print("FINAL DATASET INFORMATION")
print("========================")

print("\nTRAINING DATA")
print("Shape:", train_check.shape)
train_check.info()

print("\nVALIDATION DATA")
print("Shape:", validation_check.shape)
validation_check.info()

print("\nTEST DATA")
print("Shape:", test_check.shape)
test_check.info()
```
