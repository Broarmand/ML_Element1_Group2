```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

project_dir = Path(r"D:\MYProject\ML_Element1")

raw_path = (
    project_dir
    / "q2_own_scenario"
    / "data"
    / "raw"
    / "rwanda_traffic_accidents_synthetic.csv"
)

df = pd.read_csv(raw_path)

print("Dataset shape:", df.shape)
display(df.head())
```

```python

print("DATASET SHAPE:", df.shape)

print("\nDATA TYPES")
print(df.dtypes)

print("\nDATASET INFO")
df.info()

print("\nSUMMARY STATISTICS")
display(df.describe(include="all").T)
```

```python

missing = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (
        missing / len(df) * 100
    ).round(2)
})

display(
    missing_report[
        missing_report["Missing_Count"] > 0
    ]
)
```

```python

missing_report["Missing_Count"].sort_values().plot(
    kind="barh",
    figsize=(10, 5)
)

plt.title("Missing Values by Column")
plt.xlabel("Number of Missing Values")
plt.ylabel("Column")
plt.tight_layout()
plt.show()
```

```python

print("Exact duplicate rows:", df.duplicated().sum())

print(
    "Duplicate accident IDs:",
    df["accident_id"].duplicated().sum()
)

display(df[df.duplicated(keep=False)].head(20))
```

```python

categorical_columns = df.select_dtypes(
    include=["object", "str"]
).columns

for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False))
```

```python

numerical_columns = df.select_dtypes(
    include=["number"]
).columns

outlier_report = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_report.append({
        "Column": column,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_report)

display(outlier_report)
```

```python

fig, axes = plt.subplots(
    1, 2, figsize=(14, 5)
)

sns.histplot(
    data=df,
    x="speed_limit_kmh",
    kde=True,
    ax=axes[0]
)

axes[0].set_title("Speed Limit Distribution")

sns.histplot(
    data=df,
    x="response_time_min",
    kde=True,
    ax=axes[1]
)

axes[1].set_title("Emergency Response Time Distribution")

plt.tight_layout()
plt.show()
```

```python

target_counts = df["severity"].value_counts()

display(target_counts)

target_counts.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Distribution of Accident Severity")
plt.xlabel("Severity")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

df["district"].value_counts().plot(
    kind="bar",
    figsize=(12, 5)
)

plt.title("Simulated Accident Records by District")
plt.xlabel("District")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

severity_road = pd.crosstab(
    df["road_type"],
    df["severity"]
)

display(severity_road)

severity_road.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 5)
)

plt.title("Accident Severity by Road Type")
plt.xlabel("Road Type")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

```python

numeric_df = df.select_dtypes(
    include=["number"]
)

correlation = numeric_df.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python

quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Count": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2).values,
    "Unique_Values": df.nunique().values
})

quality_report.to_csv(
    project_dir
    / "q2_own_scenario"
    / "docs"
    / "data_quality_report.csv",
    index=False
)

display(quality_report)
```
